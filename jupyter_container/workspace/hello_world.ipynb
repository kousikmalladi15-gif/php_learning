```php
<?php

$host = "host.docker.internal";
$user_name = "root";
$pass = '';
$charset = 'utf8mb4';
$db = 'user_db';

$dsn = "mysql:host=$host;dbname=$db;charset=$charset";

$options = [
    PDO::ATTR_ERRMODE            => PDO::ERRMODE_EXCEPTION,
    PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC
];

try {
    $pdo = new PDO($dsn, $user_name, $pass, $options);
    echo "connection with $db established successfully";
} catch (\PDOException $E) {
    echo "The connection with the database failed: " . $E->getMessage();
}

?>
```
